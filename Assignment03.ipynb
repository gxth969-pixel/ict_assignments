## 2. Write a Python function to sum all the numbers in a list.

```python
def sum(numbers):
    total = 0
    for num in numbers:
        total += num
    return total
```

```python
sum([2,6,8,10])
```

# 3. Python program to insert a number to any position in a list (Get the number
to be inserted and position from the user)

```python

numbers = [10, 20, 30, 40, 50]
print("Original list:", numbers)
num = int(input("Enter the number to insert: "))
pos = int(input("Enter the position (index) to insert at: "))
numbers.insert(pos, num)

print("Updated list:", numbers)
```

# 4. Write a program to rename a key city to a location in the following dictionary.

```python
sample_dict = {
    "name": "Kelly",
    "age": 25,
    "salary": 8000,
    "city": "New york"
}

sample_dict["location"] = sample_dict.pop("city")
print("Updated dictionary:", sample_dict)
```

# 5. write a program to change Brad's salary to 85000

```python
sample_dict = {
    'emp1': {'name': 'Jhon', 'salary': 7500},
    'emp2': {'name': 'Emma', 'salary': 8000},
    'emp3': {'name': 'Brad', 'salary': 500}
}

# Update Brad's salary
sample_dict['emp3']['salary'] = 8500

# Print the updated dictionary
print(sample_dict)
```
