# 10. Develop a Log File Analyzer for simulated system logs. The program should parse text-based log entries, categorize events, count error types, analyze user activity, and filter logs based on time or severity. String splitting, pattern matching, and dictionary-based aggregation must be used. The application should present analytical summaries via a menu-driven interface.

**Algorithm: Log File Analyzer**
1. Start Program
2. Initialize Data :
   Create a list of log entries, where each entry is a text string
3. Define Log Parsing Function

   Function parse_log_entry(entry)

   Split the entry into parts using "|".

   Trim spaces from each part.

   Extract:

   *Timestamp → convert to datetime format

   *Severity

   *Username → split on "="

   *Message text

   *Return these values.
4. Define Analysis Functions

   A. Count Events by Severity

   B. Count Error Types

   C. Analyze User Activity

   D. Filter Logs by Severity

   E. Filter Logs by Time Range

5. Display Menu

   Show user options:

   1 — Show severity statistics

   
   2 — Show error type counts
   
   3 — Show user activity
   
   4 — Filter by severity
   
   5 — Filter by time
   
   6 — Show all logs
   
   0 — Exit
6. Main Menu Loop

   Repeat Until User Exits
7. End Program


```python
import re
from datetime import datetime

# -------------------------
# Simulated System Log Data
# -------------------------
logs = [
    "2026-01-01 10:15:32 | INFO  | user=john   | Login successful",
    "2026-01-01 10:17:45 | ERROR | user=john   | File not found",
    "2026-01-01 10:20:10 | WARNING | user=sarah | Disk space low",
    "2026-01-01 10:25:02 | INFO  | user=sarah | File uploaded",
    "2026-01-01 10:30:00 | ERROR | user=john   | Permission denied",
    "2026-01-01 10:40:10 | INFO  | user=admin  | System scan started",
    "2026-01-01 10:45:59 | CRITICAL | user=admin | System failure",
    "2026-01-01 11:00:05 | INFO  | user=john   | Logout successful"
]

# -------------------------
# Helper Parsing Function
# -------------------------
def parse_log_entry(entry):
    parts = [p.strip() for p in entry.split("|")]
    timestamp = datetime.strptime(parts[0], "%Y-%m-%d %H:%M:%S")
    severity = parts[1]
    user = parts[2].split("=")[1]
    message = parts[3]
    return timestamp, severity, user, message

# -------------------------
# Analytics
# -------------------------
def count_by_severity():
    counts = {}
    for entry in logs:
        _, severity, _, _ = parse_log_entry(entry)
        counts[severity] = counts.get(severity, 0) + 1
    return counts


def count_error_types():
    errors = {}
    for entry in logs:
        _, severity, _, message = parse_log_entry(entry)
        if severity in ["ERROR", "CRITICAL", "WARNING"]:
            errors[message] = errors.get(message, 0) + 1
    return errors


def user_activity():
    users = {}
    for entry in logs:
        _, _, user, _ = parse_log_entry(entry)
        users[user] = users.get(user, 0) + 1
    return users


def filter_by_severity(level):
    return [e for e in logs if f"| {level}" in e]


def filter_by_time(start, end):
    filtered = []
    for entry in logs:
        timestamp, _, _, _ = parse_log_entry(entry)
        if start <= timestamp <= end:
            filtered.append(entry)
    return filtered


# -------------------------
# Menu System
# -------------------------
def print_menu():
    print("\n===== LOG FILE ANALYZER =====")
    print("1. Show severity statistics")
    print("2. Show error type counts")
    print("3. Show user activity")
    print("4. Filter logs by severity")
    print("5. Filter logs by time range")
    print("6. Show all logs")
    print("0. Exit")
    print("=============================")


def main():
    while True:
        print_menu()
        choice = input("Select an option: ")

        if choice == "1":
            counts = count_by_severity()
            print("\nSeverity Summary:")
            for k, v in counts.items():
                print(f"{k}: {v}")

        elif choice == "2":
            errors = count_error_types()
            print("\nError Types:")
            for k, v in errors.items():
                print(f"{k} — {v}")

        elif choice == "3":
            users = user_activity()
            print("\nUser Activity:")
            for k, v in users.items():
                print(f"{k}: {v} events")

        elif choice == "4":
            level = input("Enter severity (INFO, WARNING, ERROR, CRITICAL): ").upper()
            result = filter_by_severity(level)
            print("\nFiltered Logs:")
            for r in result:
                print(r)

        elif choice == "5":
            print("Enter time as YYYY-MM-DD HH:MM:SS")
            start = datetime.strptime(input("Start time: "), "%Y-%m-%d %H:%M:%S")
            end = datetime.strptime(input("End time: "), "%Y-%m-%d %H:%M:%S")
            result = filter_by_time(start, end)
            print("\nFiltered Logs:")
            for r in result:
                print(r)

        elif choice == "6":
            print("\nAll Logs:")
            for l in logs:
                print(l)

        elif choice == "0":
            print("Exiting...")
            break

        else:
            print("Invalid selection. Try again.")


if __name__ == "__main__":
    main()
```

# 5. Build a Hostel Room Allocation and Management System. The program should manage rooms with defined capacities and student registrations. Students can be allocated rooms based on availability and constraints such as year or gender (optional). The system must ensure one student occupies only one room at a time. Vacating rooms, viewing occupancy status, and generating allocation reports should be supported.

1. Start
2. Initialize

   Create empty Room list

   Create empty Student list

3. Register Room

   Input Room details

  Store Room with capacity and constraints

4. Register Student

   Input Student details

  Store Student with no room assigned

5. Allocate Room

   Check student exists

   Check room exists

   Check student not already allocated

   Check room not full

   Check constraints (gender/year)

  Assign student to room

6. Vacate Room

  Check student exists

  Remove student from room

  Set room assigned to None

7. View Status / Reports

  Display room occupancy

  Display student allocation

8. Stop

```python
class Student:
    def __init__(self, student_id, name, year=None, gender=None):
        self.student_id = student_id
        self.name = name
        self.year = year
        self.gender = gender
        self.room_id = None


class Room:
    def __init__(self, room_id, capacity, gender=None, year=None):
        self.room_id = room_id
        self.capacity = capacity
        self.gender = gender          # Optional constraint
        self.year = year              # Optional constraint
        self.occupants = []

    def has_space(self):
        return len(self.occupants) < self.capacity


class HostelSystem:
    def __init__(self):
        self.rooms = {}
        self.students = {}

    # ---------- ROOM MANAGEMENT ----------
    def add_room(self, room_id, capacity, gender=None, year=None):
        if room_id in self.rooms:
            print("Room already exists.")
            return
        self.rooms[room_id] = Room(room_id, capacity, gender, year)
        print(f"Room {room_id} added.")

    # ---------- STUDENT MANAGEMENT ----------
    def register_student(self, student_id, name, year=None, gender=None):
        if student_id in self.students:
            print("Student already registered.")
            return
        self.students[student_id] = Student(student_id, name, year, gender)
        print(f"Student {name} registered.")

    # ---------- ALLOCATION ----------
    def allocate_room(self, student_id, room_id):
        if student_id not in self.students:
            print("Student not found.")
            return
        if room_id not in self.rooms:
            print("Room not found.")
            return

        student = self.students[student_id]
        room = self.rooms[room_id]

        if student.room_id is not None:
            print("Student already allocated a room.")
            return

        # Check constraints
        if room.gender and student.gender and room.gender != student.gender:
            print("Gender constraint prevents allocation.")
            return
        if room.year and student.year and room.year != student.year:
            print("Year constraint prevents allocation.")
            return

        if not room.has_space():
            print("Room is full.")
            return

        room.occupants.append(student_id)
        student.room_id = room_id
        print(f"{student.name} allocated to Room {room_id}.")

    # ---------- VACATE ----------
    def vacate_room(self, student_id):
        if student_id not in self.students:
            print("Student not found.")
            return

        student = self.students[student_id]

        if student.room_id is None:
            print("Student not allocated to any room.")
            return

        room = self.rooms[student.room_id]
        room.occupants.remove(student_id)
        student.room_id = None
        print(f"{student.name} has vacated Room {room.room_id}.")

    # ---------- STATUS ----------
    def view_room_status(self):
        for room in self.rooms.values():
            print(f"\nRoom {room.room_id} (Capacity {room.capacity})")
            print(f"Occupants: {len(room.occupants)}")
            for sid in room.occupants:
                s = self.students[sid]
                print(f" - {s.name} ({s.student_id})")

    def view_student_allocation(self):
        for student in self.students.values():
            print(f"{student.name} ({student.student_id}) -> {student.room_id or 'Not Allocated'}")

    # ---------- REPORT ----------
    def generate_report(self):
        print("\n=== HOSTEL ALLOCATION REPORT ===")
        total_rooms = len(self.rooms)
        total_students = len(self.students)
        allocated = sum(1 for s in self.students.values() if s.room_id)

        print(f"Total Rooms: {total_rooms}")
        print(f"Total Students: {total_students}")
        print(f"Allocated Students: {allocated}")
        print(f"Unallocated Students: {total_students - allocated}")


# ---------- DEMO ----------
if __name__ == "__main__":
    system = HostelSystem()

    # Add rooms
    system.add_room("R101", 2, gender="Male")
    system.add_room("R102", 2, gender="Female")
    system.add_room("R201", 3)

    # Register students
    system.register_student("S1", "John", year=2, gender="Male")
    system.register_student("S2", "Mary", year=1, gender="Female")
    system.register_student("S3", "Alex", year=2, gender="Male")
    system.register_student("S4", "Sophia", gender="Female")

    # Allocate
    system.allocate_room("S1", "R101")
    system.allocate_room("S3", "R101")
    system.allocate_room("S2", "R102")

    # View status
    system.view_room_status()
    system.view_student_allocation()

    # Vacate
    system.vacate_room("S1")

    # Report
    system.generate_report()
```
